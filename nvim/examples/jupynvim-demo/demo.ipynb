## DataFrame output

This is intentionally wider than a trivial table so you can judge terminal readability.

```python
rng = np.random.default_rng(42)
days = pd.date_range("2026-07-01", periods=14, freq="D")
df = pd.DataFrame({
    "day": days,
    "requests": rng.integers(800, 1600, len(days)),
    "latency_ms": rng.normal(125, 18, len(days)).round(1),
    "error_rate": rng.uniform(0.002, 0.025, len(days)).round(4),
})
df
```

```python
fig, ax = plt.subplots(figsize=(8, 3.5))
ax.plot(df["day"], df["latency_ms"], marker="o", color="#2a9d8f")
ax.axhline(140, color="#e76f51", linestyle="--", label="SLO threshold")
ax.set(title="Daily API latency", ylabel="milliseconds")
ax.grid(alpha=0.25)
ax.legend()
fig.autofmt_xdate()
plt.show()
```

```python
class ExperimentResult:
    """Small runtime-defined object for completion and hover tests."""

    def __init__(self, frame):
        self.rows = len(frame)
        self.mean = float(frame["latency_ms"].mean())
        self.maximum = float(frame["latency_ms"].max())

result = ExperimentResult(df)
vars(result)
```
